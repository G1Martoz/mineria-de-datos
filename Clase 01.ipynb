```python
# 🛠️Carga y Preprocesamiento (Limpieza)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# 1. Preparación de datos (Fase de Preprocesamiento KDD)
datos = {
    "Cliente": [1, 2, 3, 4, 5],
    "Edad": [25, 40, 32, 22, 35], # Corregido de 351 a 35 [cite: 138, 153]
    "Plan": ["Básico", "Premium", "Estándar", "Básico", "Premium"],
    "Consumo_GB": [5, 50, 10, 7, 45],
    "Reclamos": [1, 2, 3, 0, 1],
    "Estado_Cuenta": ["Pagado", "Pagado", "Moroso", "Moroso", "Pagado"] # Normalizado [cite: 138]
}

df = pd.DataFrame(datos)

# 2. Visualización (Análisis Exploratorio - EDA)
# Gráfico de barras para ver la morosidad
sns.countplot(x="Estado_Cuenta", data=df, palette="Set2")
plt.title("Distribución de Clientes por Estado de Cuenta")
plt.show()
```

```python
# 📈Análisis Exploratorio de Datos (EDA)
# Configuración de estilo
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 5))

# Histograma de Consumo de Datos [cite: 170, 229]
plt.subplot(1, 2, 1)
sns.histplot(df["Consumo_GB"], bins=5, kde=True, color="skyblue")
plt.title("Distribución de Consumo de Datos (GB)")
plt.xlabel("GB Consumidos")

# Gráfico de Barras: Estado de Cuenta [cite: 230, 256]
plt.subplot(1, 2, 2)
sns.countplot(x="Estado_Cuenta", data=df, palette="viridis")
plt.title("Frecuencia de Clientes Pagadores vs Morosos")
plt.xlabel("Estado de Cuenta")

plt.tight_layout()
plt.show()
```
