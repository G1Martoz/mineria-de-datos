Parte 2: Matplotlib y Seaborn

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Gráfico de barras (Matplotlib)
df_empleados = pd.DataFrame({
    "Categoría": ["Junior", "Semi-Senior", "Senior"],
    "Salario_Promedio": [45000, 75000, 120000]
})

plt.bar(df_empleados["Categoría"], df_empleados["Salario_Promedio"], color=["blue", "green", "red"])
plt.xlabel("Categoría de Empleado")
plt.ylabel("Salario Promedio")
plt.title("Salario Promedio por Categoría")
plt.show()

# 2. Scatterplot (Seaborn)
df2 = pd.DataFrame({
    "Edad": [25, 32, 40, 35, 50, 45, 29, 37, 41, 60],
    "Ingresos": [30000, 45000, 60000, 52000, 95000, 80000, 35000, 58000, 62000, 110000]
})

sns.scatterplot(x=df2["Edad"], y=df2["Ingresos"])
plt.title("Relación entre Edad e Ingresos")
plt.show()
```

Seaborn es más intuitivo y requiere menos código para obtener visualizaciones atractivas por defecto, mientras ue Matplotlib ofrece un control total sobre cada detalle del gráfico.

Parte 3- Graficos Avanzados (Dataset Diamonds)

```python
# Cargar dataset de diamantes
df_diamonds = sns.load_dataset("diamonds")

# 1. Pairplot (relación entre precio, quilates y profundidad)
sns.pairplot(df_diamonds[["price", "carat", "depth"]])
plt.show()

# 2. Heatmap (correlaciones)
correlaciones = df_diamonds.corr(numeric_only=True)
sns.heatmap(correlaciones, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Mapa de Correlación de Diamantes")
plt.show()

# 3. Violin Plot (distribución del precio por corte)
sns.violinplot(x="cut", y="price", data=df_diamonds, palette="muted")
plt.title("Distribución del Precio por Tipo de Corte")
plt.show()
```

Parte 4: Actividad Obligatoria (Ejercicio Integrador)

```python
# Crear DataFrame de ventas
df = pd.DataFrame({
    "Cliente": [1, 2, 3, 4, 5, 6, 7, 8],
    "Edad": [25, 40, 35, 50, 30, 45, 38, 29],
    "Ingresos": [100, 500, 300, 400, 200, 600, 450, 350],
    "Compras": [5, 15, 10, 20, 7, 18, 12, 6],
    "Categoría": ["A", "B", "C", "A", "B", "C", "A", "B"]
})

# 1) Análisis Exploratorio (EDA)
sns.histplot(df["Edad"], bins=5, kde=True)
plt.title("Distribución de Edades")
plt.show()

plt.figure(figsize=(6, 6))
sns.boxplot(y=df["Compras"], color="skyblue")
plt.grid(axis='y', linestyle='--', alpha=0.7) # Cuadrícula para medir mejor
plt.title("Boxplot de Compras V")
plt.ylabel("Cantidad de Compras")
plt.show()

sns.scatterplot(x=df["Ingresos"], y=df["Compras"])
plt.title("Relación entre Ingresos y Compras")
plt.show()

# 2) Visualización Avanzada
sns.pairplot(df[["Edad", "Ingresos", "Compras"]])
plt.show()

correlaciones = df.corr(numeric_only=True)
sns.heatmap(correlaciones, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Mapa de Correlación de Variables")
plt.show()

plt.figure(figsize=(8, 6))
# inner='quart' añade las líneas de los cuartiles para ver la métrica interna
sns.violinplot(x="Categoría", y="Compras", data=df, hue="Categoría", palette="muted", legend=False, inner='quart') 
plt.grid(axis='y', linestyle='--', alpha=0.7) # Ayuda a ver el "espacio"
plt.title("Distribución de Compras por Categoría")
plt.show()
```
